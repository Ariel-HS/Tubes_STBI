# Retrieval Benchmark — Original vs GAN-Expanded

MAP across all **32 scoring variations** (4 preprocessing × 4 TF weights × 2 normalization) on the CISI collection, evaluated over the 76 queries with result in `qrels.text`.

```python
import pandas as pd
from benchmark import run_benchmark

df = run_benchmark(progress=print)
df.to_csv("benchmark_results.csv", index=False)
print(f"\nDone — {len(df)} configs.")
```

## Full results

Sorted by expanded MAP. Greener = higher MAP; the bar shows how much expansion improved each config (`delta = map_expanded - map_original`).

```python
ranked = df.sort_values("map_expanded", ascending=False).reset_index(drop=True)

(
    ranked.style
    .format({"map_original": "{:.4f}", "map_expanded": "{:.4f}", "delta": "{:+.4f}"})
    .background_gradient(subset=["map_original", "map_expanded"], cmap="Greens")
    .bar(subset=["delta"], color="#60a5fa", align="left")
    .set_caption("MAP by configuration (sorted by expanded MAP)")
    .hide(axis="index")
)
```
